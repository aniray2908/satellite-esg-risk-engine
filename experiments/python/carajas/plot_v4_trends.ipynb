```python
import pandas as pd
import matplotlib.pyplot as plt

# Path to CSV in Google Drive
csv_path = "/content/drive/MyDrive/carajas_pit_centered_full_year_v2_1.csv"

# Load CSV
df = pd.read_csv(csv_path)

# Sort by year
df = df.sort_values("year")

# -----------------------------
# Plot 1: Mean NDVI
# -----------------------------
plt.figure()
plt.plot(df["year"], df["mean_ndvi"])
plt.title("Carajás Mine — Mean NDVI (2019–2023)")
plt.xlabel("Year")
plt.ylabel("Mean NDVI")
plt.savefig("carajas_mean_ndvi.png", dpi=300, bbox_inches="tight")
plt.show()

# -----------------------------
# Plot 2: Low NDVI Fraction
# -----------------------------
plt.figure()
plt.plot(df["year"], df["low_ndvi_fraction"])
plt.title("Carajás Mine — Low NDVI Fraction (NDVI < 0.2)")
plt.xlabel("Year")
plt.ylabel("Low NDVI Fraction")
plt.savefig("carajas_low_ndvi_fraction.png", dpi=300, bbox_inches="tight")
plt.show()
```
