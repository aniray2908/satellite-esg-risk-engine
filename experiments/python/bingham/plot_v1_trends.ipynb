```python
import pandas as pd
import matplotlib.pyplot as plt

# Update path if needed
csv_path = "/content/drive/MyDrive/bingham_pit_centered_full_year_v1.csv"

# Load data
df = pd.read_csv(csv_path)
df = df.sort_values("year")

# -----------------------------
# Plot 1 — Mean NDVI
# -----------------------------
plt.figure()
plt.plot(df["year"], df["mean_ndvi"])
plt.title("Bingham Canyon — Mean NDVI (2019–2023)")
plt.xlabel("Year")
plt.ylabel("Mean NDVI")
plt.savefig("bingham_mean_ndvi.png", dpi=300, bbox_inches="tight")
plt.show()

# -----------------------------
# Plot 2 — Low NDVI Fraction
# -----------------------------
plt.figure()
plt.plot(df["year"], df["low_ndvi_fraction"])
plt.title("Bingham Canyon — Low NDVI Fraction (NDVI < 0.2)")
plt.xlabel("Year")
plt.ylabel("Low NDVI Fraction")
plt.savefig("bingham_low_ndvi_fraction.png", dpi=300, bbox_inches="tight")
plt.show()
```
